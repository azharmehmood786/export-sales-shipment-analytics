```python
import pandas as pd
from datetime import date, timedelta
```

```python
products=pd.DataFrame([{"product_id": 1, "product_name": "Rice", "category": "Food", "unit_price_usd": 20},
    {"product_id": 2, "product_name": "Cotton", "category": "Textile", "unit_price_usd": 30},
    {"product_id": 3, "product_name": "Sports Goods", "category": "Sports", "unit_price_usd": 45},
    {"product_id": 4, "product_name": "Leather Bags", "category": "Leather", "unit_price_usd": 35}])
```

```python
customers = [
    {
        "customer_id": i,
        "customer_name": f"Customer_{i}",
        "country": ["USA", "UK", "Germany", "UAE", "China"][(i-1) % 5]
    }
    for i in range(1, 21)
]

customers = pd.DataFrame(customers)

customers.head(20)
```

```python
print(type(products))
print(type(customers))
```

```python
import pandas as pd
from datetime import date, timedelta

# Generate 500 orders
orders_data = []

for i in range(1, 501):

    product = products.iloc[(i - 1) % len(products)]
    customer = customers.iloc[(i - 1) % len(customers)]

    order_date = date(2026, 1, 1) + timedelta(days=(i - 1) % 300)
    promised_date = order_date + timedelta(days=10)
    actual_date = promised_date + timedelta(days=(i % 5) - 2)

    quantity = 10 + (i % 100)

    orders_data.append({
        "order_id": i,
        "customer_id": customer["customer_id"],
        "product_id": product["product_id"],
        "order_date": order_date,
        "promised_ship_date": promised_date,
        "actual_ship_date": actual_date,
        "quantity": quantity,
        "unit_price_usd": product["unit_price_usd"],
        "total_amount_usd": quantity * product["unit_price_usd"],
        "order_type": "Export" if i % 2 == 0 else "Domestic"
    })

# Convert list into DataFrame
orders_df = pd.DataFrame(orders_data)

# Display result
print(orders_df.shape)
display(orders_df.head())
```

```python
import os

# Create data folder
os.makedirs("data", exist_ok=True)
```

```python
products.to_csv("data/products.csv", index=False)
customers.to_csv("data/customers.csv", index=False)
orders_df.to_csv("data/orders.csv", index=False)

print("All CSV files saved successfully!")
```

```python
import sqlite3

# Create database
conn = sqlite3.connect("export_business.db")
```

```python
products.to_sql("products", conn, if_exists="replace", index=False)

customers.to_sql("customers", conn, if_exists="replace", index=False)

orders_df.to_sql("orders", conn, if_exists="replace", index=False)

print("All tables loaded successfully!")
```

```python
cursor = conn.cursor()

cursor.execute("""
SELECT name FROM sqlite_master
WHERE type='table';
""")

print(cursor.fetchall())
```

```python
query="""
SELECT SUM(quantity) AS total_export_volume
FROM orders
WHERE order_type = 'Export';
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    ROUND(
        SUM(CASE
            WHEN actual_ship_date <= promised_ship_date
            THEN 1 ELSE 0
        END) * 100.0 / COUNT(*), 2
    ) AS on_time_shipment_rate
FROM orders;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT AVG(total_amount_usd) AS average_order_value
FROM orders;
"""

pd.read_sql_query(query, conn)
```
